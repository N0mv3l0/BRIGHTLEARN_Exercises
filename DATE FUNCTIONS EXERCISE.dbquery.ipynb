```sql
CREATE DATABASE DATE_FUNCTIONS;

USE DATE_FUNCTIONS;

-- TABLE 1
CREATE TABLE orders(
    order_id INT,
    customer_id INT,
    order_date DATE
);

INSERT INTO orders
VALUES
('1001','101','2026-05-01'),
('1002','102','2026-05-02'),
('1003','103','2026-05-03'),
('1004','104','2026-05-04'),
('1005','105','2026-05-05');

SELECT *
FROM orders;

-- TABLE 02
CREATE TABLE customer_signups(
    customer_id INT,
    customer_name STRING,
    signup_date DATE
);

INSERT INTO customer_signups
VALUES
('201','John','2026-01-15'),
('202','Mary','2026-02-20'),
('203','Peter','2026-03-05'),
('204','Sarah','2026-04-18'),
('205','Thabo','2026-05-30');

select *
from customer_signups;

--TABLE 03
CREATE TABLE sales(
    sale_id STRING,
    product_name STRING,
    sale_date DATE,
    amount INT
);

INSERT INTO sales
VALUES
('S001','Laptop','2026-01-10','12000'),
('S002','Mouse','2026-02-15','250'),
('S003','Keyboard','2026-03-20','700'),
('S004','Monitor','2026-04-25','3500'),
('S005','Desk','2026-05-30','2000');

SELECT *
FROM sales;

-- TABLE 04
CREATE TABLE transactions(
    transaction_id STRING,
    customer_id INT,
    transaction_date DATE,
    amount INT
);

INSERT INTO transactions
VALUES
('T001','101','2024-12-15','500'),
('T002','102','2025-01-20','1200'),
('T003','103','2025-06-10','800'),
('T004','104','2026-02-05','1500'),
('T005','105','2026-05-25','2000');

SELECT *
FROM transactions;

-- TABLE 05
CREATE TABLE deliveries(
    delivery_id STRING,
    customer_id INT,
    delivery_date DATE
);

INSERT INTO deliveries
VALUES
('D001','101','2026-05-01'),
('D002','102','2026-05-08'),
('D003','103','2026-05-15'),
('D004','104','2026-05-22'),
('D005','105','2026-05-29');

SELECT *
FROM deliveries;

--TABLE 06
CREATE TABLE employees(
    employee_id INT,
    employee_name STRING,
    department STRING
); 

INSERT INTO employees
VALUES
('301','Nandi','Sales'),
('302','Brian','IT'),
('303','Lerato','Finance'),
('304','Sipho','HR'),
('305','Aisha','Marketing');

SELECT *
FROM employees;

-- TABLE 07
CREATE TABLE online_order(
    order_id INT,
    customer_id INT,
    order_date_text DATE
);

INSERT INTO online_order
VALUES
('4001','101','2026-01-15'),
('4002','102','2026-02-20'),
('4003','103','2026-03-25'),
('4004','104','2026-04-10'),
('4005','105','2026-05-05');

SELECT *
FROM online_order;

--Table 08
CREATE TABLE payment_dates(
    payment_id STRING,
    customer_id INT,
    payment_date DATE
);

INSERT INTO payment_dates
VALUES
('P001','101','2026-01-05'),
('P002','102','2026-02-10'),
('P003','103','2026-03-15'),
('P004','104','2026-04-20'),
('P005','105','2026-05-25');

SELECT *
FROM payment_dates;

-- TABLE 09
CREATE TABLE customer_purchases(
    customer_id INT,
    customer_name STRING,
    last_purchase_date DATE
);

INSERT INTO customer_purchases
VALUES
('501','John','2026-05-01'),
('502','Mary','2026-05-10'),
('503','Peter','2026-05-15'),
('504','Sarah','2026-05-20'),
('505','Thabo','2026-05-25');

SELECT *
FROM customer_purchases;

-- TABLE 10
CREATE TABLE shipping_orders(
    order_id INT,
    customer_id INT,
    order_date DATE
);

INSERT INTO shipping_orders
VALUES
('6001','101','2026-05-01'),
('6002','102','2026-05-03'),
('6003','103','2026-05-05'),
('6004','104','2026-05-07'),
('6005','105','2026-05-09');

SELECT *
FROM shipping_orders;

--TABLE 11
CREATE TABLE bookings(
    booking_id STRING,
    customer_id INT,
    booking_date DATE
);

INSERT INTO bookings
VALUES
('B001','101','2026-01-12'),
('B002','102','2026-02-18'),
('B003','103','2026-03-22'),
('B004','104','2026-04-09'),
('B005','105','2026-05-27');

SELECT *
FROM bookings;

-- TABLE 12
CREATE TABLE yearly_orders(
    order_id INT,
    customer_id INT,
    order_date DATE,
    amount INT
);

INSERT INTO yearly_orders
VALUES
('7001','101','2024-12-15','500'),
('7002','102','2025-01-20','1200'),
('7003','103','2025-06-10','800'),
('7004','104','2026-02-05','1500'),
('7005','105','2026-05-25','2000');

SELECT *
FROM yearly_orders;

-- TABLE 13
CREATE TABLE monthly_orders(
    order_id INT,
    customer_id INT,
    order_date DATE,
    amount INT
);

INSERT INTO monthly_orders
VALUES
('8001','101','2026-01-15','500'),
('8002','102','2026-02-20','1200'),
('8003','103','2026-03-10','800'),
('8004','104','2026-03-25','1500'),
('8005','105','2026-05-30','2000');

SELECT *
FROM monthly_orders;

-- TABLE 14
CREATE TABLE subscriptions(
    subscription_id STRING,
    customer_id INT,
    start_date DATE
);

INSERT INTO subscriptions
VALUES
('SUB001','101','2026-01-10'),
('SUB002','102','2026-02-15'),
('SUB003','103','2026-03-20'),
('SUB004','104','2026-04-25'),
('SUB005','105','2026-05-30');

SELECT *
FROM subscriptions;

--TABLE 15
CREATE TABLE campaign_sends(
    send_id STRING,
    customer_id INT,
    send_date DATE
);

INSERT INTO campaign_sends
VALUES
('C001','101','2026-01-12'),
('C002','102','2026-02-18'),
('C003','103','2026-03-22'),
('C004','104','2026-04-09'),
('C005','105','2026-05-27');

SELECT *
FROM campaign_sends;

-- TABLE 16
CREATE TABLE invoice_dates(
    invoice_id STRING,
    customer_id INT,
    invoice_date DATE
);

INSERT INTO invoice_dates
values
('INV001','101','2026-01-05'),
('INV002','102','2026-02-10'),
('INV003','103','2026-03-15'),
('INV004','104','2026-04-20'),
('INV005','105','2026-05-25');

SELECT *
FROM invoice_dates;

-- TABLE 17
CREATE TABLE customer_birthdays(
    customer_id INT,
    customer_name STRING,
    date_of_birth DATE
);

INSERT INTO customer_birthdays
VALUES
('901','John','1998-05-10'),
('902','Mary','1990-08-20'),
('903','Peter','2002-03-15'),
('904','Sarah','1985-12-01'),
('905','Thabo','2000-07-30');

SELECT *
FROM customer_birthdays;

-- TABLE 18
CREATE TABLE weekend_orders(
    order_id INT,
    customer_id INT,
    order_date DATE
);

INSERT INTO weekend_orders
VALUES
('9001','101','2026-05-01'),
('9002','102','2026-05-02'),
('9003','103','2026-05-03'),
('9004','104','2026-05-04'),
('9005','105','2026-05-05');

SELECT * 
FROM weekend_orders;

-- TABLE 19
CREATE OR REPLACE TABLE quarterly_transactions(
    transaction_id STRING,
    customer_id INT,
    transaction_date DATE,
    amount INT);

 
INSERT INTO quarterly_transactions
VALUES
('Q001','101','2026-01-15','500'),
('Q002','102','2026-03-20','1200'),
('Q003','103','2026-04-10','800'),
('Q004','104','2026-07-05','1500'),
('Q005','105','2026-10-25','2000');

SELECT *
FROM quarterly_transactions;

-- TABLE 20
CREATE TABLE recent_orders(
    order_id STRING,
    customer_id INT,
    order_date DATE,
    amount INT
);

INSERT INTO recent_orders
VALUES
('R001','101','2026-04-01','500'),
('R002','102','2026-04-15','1200'),
('R003','103','2026-05-01','800'),
('R004','104','2026-05-10','1500'),
('R005','105','2026-05-25','2000');

SELECT *
FROM recent_orders;

-- TABLE 21- BONUS CHALLENGE
CREATE TABLE customer_recency(
    customer_id INT,
    customer_name STRING,
    last_purchase_date DATE,
    total_spend INT
);

INSERT INTO workspace.date_functions.customer_recency
VALUES
('1001','John','2026-05-25','5000'),
('1002','Mary','2026-05-10','2500'),
('1003','Peter','2026-04-01','700'),
('1004','Sarah','2026-02-15','15000'),
('1005','Thabo','2025-12-20','300');

SELECT *
FROM workspace.date_functions.customer_recency;

---- SOLUTIONS

1. Return each order with the day name of the order date.

SELECT order_id, 
       customer_id, 
       order_date,
       DAYNAME(order_date) AS day_name
FROM workspace.date_functions.orders;

2. Return each customer with the name of the month in which they signed up.

SELECT customer_id, 
       customer_name, 
       signup_date, 
       MONTHNAME(signup_date) AS signup_month_name
FROM workspace.date_functions.customer_signups;

3. Return each sale with the month number extracted from sale_date.

SELECT sale_id, 
       product_name, 
       sale_date, 
       MONTH(sale_date) AS sale_month
FROM workspace.date_functions.sales;

4. Return each transaction with the year extracted from transaction_date.

SELECT transaction_id, 
       customer_id, 
       transaction_date, 
       YEAR(transaction_date) AS transaction_year
FROM workspace.date_functions.transactions;

5. Return each delivery with the day of the month extracted from delivery_date.

SELECT delivery_id, 
       customer_id, 
       delivery_date,
       DAY(delivery_date) AS day_of_month
FROM workspace.date_functions.deliveries;

6. Return all employees and add a column showing today's date.

SELECT employee_id, 
       employee_name, 
       department, 
       CURRENT_DATE() AS today_date
FROM workspace.date_functions.employees;

7. Convert the order_date_text string column into a proper date column.

SELECT order_id, 
       customer_id, 
       order_date_text, 
       date_format(order_date_text, 'yyyy-MM-dd') AS order_date
FROM workspace.date_functions.online_order;

8. Format payment_date as a text string in the format YYYY-MM-DD.

SELECT payment_id, 
       customer_id, 
       payment_date, 
       date_format(payment_date,'yyyy-MM-DD') AS formatted_payment_date
FROM workspace.date_functions.payment_dates;

9. Calculate how many days have passed since each customer's last purchase. Use today's 
date.

SELECT customer_id, 
       customer_name, 
       last_purchase_date, 
       Date_diff(current_date(),last_purchase_date) AS days_since_last_purchase
FROM workspace.date_functions.customer_purchases;

10. Calculate the expected delivery date by adding 7 days to the order_date.

SELECT order_id, 
       customer_id, 
       order_date, 
       DATE_ADD(order_date,7) AS expected_delivery_date
FROM workspace.date_functions.shipping_orders;

11. Extract the year, month number, and day from booking_date into three separate columns.

SELECT booking_id, 
       customer_id, 
       booking_date, 
       YEAR(booking_date) AS booking_year, 
       MONTH(booking_date) AS booking_month, 
       DAY(booking_date) AS booking_day
FROM workspace.date_functions.bookings;

12. Return only orders from the year 2026.
SELECT order_id, 
       customer_id, 
       order_date, 
       YEAR(order_date) AS order_year, 
       amount
FROM workspace.date_functions.yearly_orders
WHERE YEAR(order_date) = 2026;

13. Return only orders placed in March (month number 3).

SELECT order_id, 
       customer_id, 
       order_date, 
       MONTH(order_date) AS order_month,
       amount
FROM workspace.date_functions.monthly_orders
WHERE MONTH(order_date) = 3;

14. Return the last day of the month for each subscription start date.

SELECT subscription_id, 
       customer_id, 
       start_date, 
       last_day(start_date) AS month_end_date
FROM workspace.date_functions.subscriptions;

15. Return the first day of the month for each campaign send date.
SELECT send_id, 
       customer_id, 
       send_date, 
       DATE_TRUNC('MONTH',send_date) AS month_start_date
FROM workspace.date_functions.campaign_sends;

16. Format invoice_date as a text string showing month name and year. Example: January 
2026.

SELECT invoice_id, 
       customer_id, 
       invoice_date, 
       DATE_FORMAT(invoice_date,'MMMM-yyyy') AS invoice_month_year
FROM workspace.date_functions.invoice_dates;

-- To get month, use 'MM',
-- To get month name, use 'MMMM'.

17. Calculate each customer's age in years using their date of birth and today's date

SELECT customer_id, 
       customer_name, 
       date_of_birth, 
       FLOOR(DATE_DIFF(Current_date(),date_of_birth)/365) AS customer_age 
FROM workspace.date_functions.customer_birthdays; -- I learned that 'FLOOR' removes decimals, now I can get the exact age of the clients.

18. Classify each order as Weekend or Weekday based on the day name of the order date.
Saturday and Sunday = Weekend. All others = Weekday.

SELECT order_id, 
       customer_id, 
       order_date, 
       DAYNAME(order_date) AS day_name, 
    CASE 
        WHEN DAYOFWEEK(order_date) IN (1,7) THEN 'WEEKEND'
        ELSE 'WEEKDAY'
        END AS day_type
FROM workspace.date_functions.weekend_orders; 
-- the 'IN' statement filters the data for the exact items you want. Examples to filter departments only in IT and HR, nothing outside those two items must show up in the output.
--Databricks understands the number of days of weeks like so:
-- 1- Sunday, 2- Mon, 3-Tues, 4- Wed, 5- Thur, 6- Fri, and 7- Sun.
-- So I used the 'IN' statement to find days that are 1 and 7, which makes them part of the WEEKEND. 

19. Extract the quarter number from each transaction date

SELECT transaction_id, 
       customer_id, 
       transaction_date,
       Quarter(transaction_date) AS transaction_quarter, 
       amount
FROM workspace.date_functions.quarterly_transactions;

20. Return only orders that are more than 30 days old from today. Calculate days_since_order 
using today's date.

SELECT order_id, 
       customer_id, 
       order_date, 
       DATE_DIFF(current_date(),order_date) AS days_since_order, 
       amount
FROM workspace.date_functions.recent_orders
WHERE DATE_DIFF(current_date(), order_date)  > 30; -- I keep confusing myself with the single quotes ('') on the date_diff function.
-- I kept gettinf "error" because I included them on the order_date.

-- BONUS
21. Calculate days_since_last_purchase for each customer using CURRENT_DATE. Then use 
CASE to classify: ≤ 30 days = Active Customer, 31–90 days = At Risk Customer, > 90 days = 
Inactive Customer.

SELECT customer_id, 
       customer_name, 
       last_purchase_date, 
       DATE_DIFF(current_date(),last_purchase_date) AS days_since_last_purchase, 

  CASE WHEN DATE_DIFF(current_date(),last_purchase_date) <= 30 THEN 'Active Customer'
       WHEN DATE_DIFF(current_date(),last_purchase_date) BETWEEN 31 AND 90 THEN 'At Risk Customer'
       ELSE 'Inactive Customer'
       END AS customer_status
FROM workspace.date_functions.customer_recency; 
-- My errors were me inserting a comma after 'THEN'
-- Also I wrote the less than or equal to sign link (= <), but the correct sign is '<=' with so spacing inbetween.
```
