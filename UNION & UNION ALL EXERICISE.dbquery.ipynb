```sql
CREATE DATABASE UNION;


CREATE SCHEMA NEXBANK;

USE NEXBANK;

-- 20 TABLES FROM EXERCISE

-- TABLE 01
CREATE TABLE Branch_sandton_account(
    account_id STRING,
    account_holder STRING
);

INSERT INTO Branch_sandton_account
VALUES
('A001','Nomvula Dlamini'),
('A002','David Mokoena'),
('A003','Lerato Sithole'),
('A004','Peter Nkosi');

SELECT *
FROM Branch_sandton_account;

-- TABLE 02
CREATE TABLE Branch_rosebank_accounts(
    account_id STRING,
    account_holder string
);

INSERT INTO Branch_rosebank_accounts
VALUES
('A003','Lerato Sithole'),
('A004','Peter Nkosi'),
('A005','Zanele Khumalo'),
('A006','Thabo Motha');

SELECT *
FROM Branch_rosebank_accounts;

-- TABLE 03

CREATE TABLE savings_products(
    product_code STRING,
    product_name STRING
);

INSERT INTO savings_products
VALUES
('SV01','Basic Savings'),
('SV02','Premium Savings'),
('SV03','Youth Savings'),
('SV04','Business Savings');

SELECT *
FROM savings_products;

-- Table 04
CREATE TABLE current_products(
    product_code STRING,
    product_name STRING
);

INSERT INTO current_products
VALUES
('CR01','Standard Current'),
('CR02','Gold Current'),
('SV03','Youth Savings'),
('CR04','Business Current');

SELECT *
FROM current_products;

-- TABLE 05
CREATE TABLE retail_banking_staff(
    staff_id STRING,
    staff_name STRING
);

INSERT INTO retail_banking_staff
VALUES
('S001','Mpho Radebe'),
('S002','Brian Tshabalala'),
('S003','Aisha Patel'),
('S004','Kabelo Moabelo');

select *
from retail_banking_staff;

--TABLE 06

CREATE TABLE corporate_banking_staff(
    staff_id STRING,
    staff_name STRING
);

INSERT INTO corporate_banking_staff
VALUES
('S003','Aisha Patel'),
('S005','Nandi Dube'),
('S006','Sipho Khumalo'),
('S004','Kabelo Moabelo');

select *
from corporate_banking_staff;

-- TABLE 07
CREATE TABLE mobile_branch_cities(
    city_code STRING,
    city_name STRING
);

INSERT INTO mobile_branch_cities
VALUES
('C01','Johannesburg'),
('C02','Pretoria'),
('C03','Cape Town'),
('C04','Durban');

SELECT *
FROM mobile_branch_cities;

-- TABLE 08
CREATE TABLE digital_branch_cities(
    city_code STRING,
    city_name STRING
);

INSERT INTO digital_branch_cities
VALUES
('C03','Cape Town'),
('C05','Polokwane'),
('C06','Port Elizabeth'),
('C01','Johannesburg');

SELECT *
FROM digital_branch_cities;

-- TABLE 09

CREATE TABLE push_notification_targets(
    customer_id STRING,
    customer_name STRING
);

INSERT INTO push_notification_targets
VALUES
('C1001','Nomsa Zwane'),
('C1002','Andile Buthelezi'),
('C1003','Fatima Mohamed'),
('C1004','Ryno van Zyl');

SELECT *
FROM push_notification_targets;

--TABLE 10
CREATE TABLE inapp_banner_targets(
    customer_id STRING,
    customer_name STRING
);

INSERT INTO inapp_banner_targets
VALUES
('C1003','Fatima Mohamed'),
('C1005','Thandeka Cele'),
('C1006','Samuel Nkosi'),
('C1002','Andile Buthelezi');

SELECT *
FROM inapp_banner_targets;

-- TABLE 11
CREATE TABLE atm01_transactions(
    txn_id STRING,
    account_id STRING,
    amount INT
);

INSERT INTO atm01_transactions
VALUES
('T1001','A001','500'),
('T1002','A002','1200'),
('T1003','A001','300'),
('T1004','A003','750');

SELECT *
FROM atm01_transactions;

-- TABLE 12
CREATE TABLE atm02_transactions(
    txn_id STRING,
    account_id STRING,
    amount INT
);

INSERT INTO atm02_transactions
VALUES
('T1003','A001','300'),
('T1005','A004','900'),
('T1006','A002','450'),
('T1007','A005','150');

SELECT *
FROM atm02_transactions;

-- TABLE 13
CREATE TABLE gauteng_loan_applications(
    app_id STRING,
    customer_id STRING,
    loan_type STRING
);

INSERT INTO  gauteng_loan_applications
VALUES
('LA001','C1001','Home loan'),
('LA002','C1002','Vehicle loan'),
('LA003','C1003','Personal loan'),
('LA004','C1004','Home loan');

SELECT *
FROM gauteng_loan_applications;

-- TABLE 14
CREATE TABLE western_cape_loan_applications(
    app_id STRING,
    customer_id STRING,
    loan_type STRING
);

INSERT INTO western_cape_loan_applications
VALUES
('LA003','C1003','Personal loan'),
('LA005','C1005','Vehicle loan'),
('LA006','C1006','Business loan'),
('LA007','C1007','Home loan');

SELECT *
FROM western_cape_loan_applications;

-- TABLE 15
CREATE TABLE email_complaints(
    complaint_id STRING,
    customer_id STRING
);

INSERT INTO email_complaints
VALUES
('EC001','C2001'),
('EC002','C2002'),
('EC003','C2003'),
('EC004','C2004');

SELECT *
FROM email_complaints;

-- TABLE 16
CREATE TABLE app_complaints(
    complaint_id STRING,
    customer_id STRING
);

INSERT INTO app_complaints
VALUES
('AC001','C2005'),
('AC002','C2001'),
('AC003','C2006'),
('AC004','C2007');

SELECT *
FROM app_complaints;

--TABLE 17
CREATE TABLE april_payments(
    payment_id STRING,
    account_id STRING,
    amount_paid INT
);

INSERT INTO  april_payments
VALUES
('PAY001','A001','12500'),
('PAY002','A002','4800'),
('PAY003','A003','9200'),
('PAY004','A004','3300');

SELECT *
FROM april_payments;

--TABLE 18
CREATE TABLE may_payments(
    payment_id STRING,
    account_id STRING,
    amount_paid INT
);

INSERT INTO may_payments
VALUES
('PAY005','A001','12500'),
('PAY006','A005','7600'),
('PAY007','A002','5100'),
('PAY008','A006','2800');

SELECT payment_id,
        account_id,
        amount_paid
FROM may_payments;

-- TABLE 19
CREATE TABLE debit_entries(
    entry_id STRING,
    account_id STRING,
    entry STRING
);

INSERT INTO debit_entries
VALUES
('DR001','A001','Debit'),
('DR002','A002','Debit'),
('DR003','A003','Debit'),
('DR004','A004','Debit');

SELECT *
FROM debit_entries;

-- TABLE 20
CREATE TABLE credit_entries(
    entry_id STRING,
    account_id STRING,
    entry STRING
);

INSERT INTO credit_entries
VALUES
('CR001','A001','Credit'),
('CR002','A005','Credit'),
('CR003','A002','Credit'),
('CR004','A006','Credit');

SELECT *
FROM credit_entries;

-- SOLUTIONS
1. Combine the account holders from both branches into one unique list. Each person 
should appear only once, even if they hold accounts at both branches

SELECT DISTINCT account_id,  
       account_holder,
       'Sandton' AS city
FROM Branch_sandton_account
    UNION
SELECT DISTINCT account_id,  
       account_holder,
       'Rosebank' AS city
FROM Branch_rosebank_accounts;

-- 'Sandon' and 'Rosebank' are to be in single quotes because they are not columns in our dataset.
-- City does not need single quotes because it's an alias.
-- Wheather I use SELECT or SELECT DISTINCT, the outcome is the same. Lerato Sithole and Peter appear twice because the city column differs.

2. NexBank offers two product lists — one for savings accounts and one for current 
accounts. Build a unique product catalogue that shows every product the bank offers, 
without repeating any product that appears in both lists.
 
 SELECT product_code,  
        product_name,  
       'Savings' AS product_type
FROM savings_products
    UNION
SELECT product_code,  
        product_name,  
        'Current' AS product_type
FROM current_products;
-- SV03 appears twice but under different product types, because the original tables show it like that.

3. The HR team needs a unique list of all staff who work in either Retail Banking or 
Corporate Banking. Some staff are assigned to both. Each staff member should appear 
only once.

SELECT DISTINCT staff_id,  
       staff_name,  
       'retail' AS email
FROM retail_banking_staff
    UNION
SELECT DISTINCT staff_id, 
       staff_name,  
       'corporate' AS email
FROM corporate_banking_staff;

4. Marketing needs a unique list of all cities where NexBank serves customers — either 
through its mobile branches or its digital branch (which serves customers online 
regardless of city). No city should appear twice.

SELECT city_code,  
       city_name, 
       'Mobile' AS region
FROM workspace.nexbank.mobile_branch_cities
    UNION
SELECT city_code,  
       city_name,  
      'Digital' AS region
FROM workspace.nexbank.digital_branch_cities;

5. The digital marketing team ran two campaigns last month — one via push notification 
and one via in-app banner. Build a unique list of every customer who was targeted. If a 
customer received both, they must appear only once.

SELECT customer_id,  
       customer_name,  
       'push' AS segment
FROM workspace.nexbank.push_notification_targets
    UNION
SELECT customer_id,  
       customer_name,  
      'inapp' AS segment
FROM workspace.nexbank.inapp_banner_targets;

6. The fraud team needs the complete log of all ATM transactions from machine ATM-01 
and machine ATM-02. Every transaction must appear — even if the same transaction 
appears in both logs due to a system sync error

SELECT txn_id,  
       account_id,  
       amount    
FROM atm01_transactions
    UNION ALL
SELECT txn_id,  
       account_id,  
       amount   
FROM atm02_transactions;

7. The credit team wants a full list of all loan applications received from the Gauteng 
region and the Western Cape region. Every application must be included — duplicates 
are allowed because both regions may have captured the same application.

SELECT app_id,  
       customer_id,  
       loan_type
FROM gauteng_loan_applications
 UNION ALL
SELECT app_id,  
       customer_id,  
       loan_type
FROM western_cape_loan_applications;

8. The customer experience team needs a full record of every complaint logged this 
month, whether it came via email or via the mobile app. Every complaint must appear 
— the team needs all of them to calculate total complaint volumes.

SELECT complaint_id,  
       customer_id
FROM email_complaints
    UNION ALL
SELECT complaint_id, 
       customer_id
FROM app_complaints;

9. Finance needs to reconcile all payment records from April and May into one table for 
the half-year report. Every payment from both months must appear — the total row 
count matters for reconciliation.

SELECT payment_id,  
       account_id,  
       amount_paid
FROM april_payments
    UNION ALL
SELECT payment_id,  
       account_id,  
       amount_paid
FROM may_payments;

10. The accounting system records debits and credits in two separate tables. Combine all 
entries into one general ledger for the audit. Every single entry from both tables must 
appear — no rows should be dropped.

SELECT entry_id,  
       account_id,  
       entry
FROM debit_entrieS
    UNION ALL
SELECT entry_id, 
       account_id,  
       entry
FROM credit_entries;

-- BONUS CHALLENGE
1. A fraud analyst wants a unique list of customers who appeared in either the January or
February high-risk watch lists. Should they use UNION or UNION ALL? Explain why.

-- Use UNION because it removes duplicates. The same client will not appear 3 times in the same list.


2. The auditing team wants to count every individual transaction that passed through the
bank in Q1 — even if the same transaction was recorded in two systems on the same
day. Should they use UNION or UNION ALL? Explain why.

--UNION ALL because the outcome shows every transaction.

3. A developer writes this query:  SELECT account_id, account_holder FROM 
branch_sandton_accounts UNION SELECT account_id, account_holder, city FROM 
branch_rosebank_accounts;  The query fails. What is wrong, and how would you fix it?

-- The query will fail because in order to use UNION or UNION ALL, the tables you are joining should be similar.
-- The branch_sandton_accounts table has two columns ( account_id and account_holder), while the branch_rosebank_accounts, has 3 columns selected, the third one being 'city'. To fix the query, the developer should either remove 'city' from branch_rosebank_accounts, or say: 
 SELECT account_id, account_holder, 'sandton' AS city 
 FROM branch_sandton_accounts
 UNION 
 SELECT account_id, account_holder, 'rosebank' AS city
FROM 
branch_rosebank_accounts;

Then the query will run.
```
